```python
import numpy as np

def one_hot(tokens, vocab):
    vocab_index = {v:i for i,v in enumerate(vocab)}
    result = np.zeros((len(tokens), len(vocab)))
    for i, t in enumerate(tokens):
        if t in vocab_index:
            result[i, vocab_index[t]] = 1
    return result

# Test
print(one_hot(['A', 'C', 'G'], ['A','C','G','T']))
```

```python
import numpy as np

def cosine_sim(a, b):
    a = np.array(a)
    b = np.array(b)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print(cosine_sim([1,2,3], [2,3,4]))
```

```python
def rob(nums):
        rob, norob = 0, 0
        for num in nums:
            newRob = norob + num
            newNoRob = max(norob, rob)
            rob, norob = newRob, newNoRob
        return max(rob, norob)
```

```python
print(rob([5,1,1,9]))
```
