```python
import os
from huggingface_hub import hf_hub_download
from transformers import AutoTokenizer
from src.data_loader.molecule_tokenizer import MoleculeTokenizer
```

```python
# Create tokenizers directory
os.makedirs("./data/tokenizers/", exist_ok=True)

# Method 1: Download tokenizer.json directly
tokenizer_path = hf_hub_download(
    repo_id="bisectgroup/NovoMolGen_32M_SAFE_BPE",
    filename="tokenizer.json",
    local_dir="./data/tokenizers/",
    local_dir_use_symlinks=False
)

print(f"Downloaded tokenizer to: {tokenizer_path}")
```

```python
# Method 2: Use AutoTokenizer (alternative)
tokenizer = AutoTokenizer.from_pretrained("bisectgroup/NovoMolGen_32M_SAFE_BPE")
tokenizer.save_pretrained("./data/tokenizers/novomolgen_32M_safe_bpe/")
```

```python
# Test the tokenizer
tokenizer_path = "data/tokenizers/novomolgen_32M_safe_bpe/tokenizer.json"
mol_tokenizer = MoleculeTokenizer.load(tokenizer_path)
hf_tokenizer = mol_tokenizer.get_pretrained()

# Test encoding/decoding
test_smiles = "CCOCCNCC=CC=CC=C"
encoded = hf_tokenizer.encode(test_smiles)
decoded = hf_tokenizer.decode(encoded)
print(f"Original: {test_smiles}")
print(f"Encoded: {encoded}")
print(f"Decoded: {decoded}")
```
