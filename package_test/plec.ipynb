```python
import numpy as np
from unimol_tools import UniMolRepr
import torch
import oddt
import oddt as od
from oddt import fingerprints
```

```python
def calc_ifp_plec(prot_path: str, lig_path: str) -> torch.Tensor:
    try:
        import oddt
        protein = next(oddt.toolkit.readfile('pdb', prot_path))
        protein.protein = True
        ligand = next(oddt.toolkit.readfile('sdf', lig_path))
        
        ifp = od.fingerprints.PLEC(
            protein,ligand,
            depth_ligand = 1,
            depth_protein  = 5,
            distance_cutoff= 4.5,
            size           = 16384,
            count_bits     = True,
            sparse         = False,
            ignore_hoh     = True,
        )
        
        return torch.tensor(ifp, dtype=torch.float32).unsqueeze(0)
    except Exception as e:
        return None, f"{prot_path} or {lig_path}: {e}"
```
